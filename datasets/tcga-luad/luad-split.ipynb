## Preprossing TCGA-LUAD non-omics data

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder
from sklearn.impute import SimpleImputer
```

```python
data = pd.read_csv("D:/NCKH/CURE/data/tcga-luad/luad_tcga_pan_can_atlas_2018_clinical_data.tsv", sep="\t")
```

```python
data = data.rename(columns={'Sample ID': 'SAMPLE_ID'})
```

```python
data.columns
```

```python
print(data.isna().sum().sort_values(ascending=False).to_string())
```

```python
data = data.dropna(subset=['Overall Survival (Months)'])
```

```python
print(data.isna().sum().sort_values(ascending=False).to_string())
```

```python
drop_cols = ["Patient Weight", "Primary Lymph Node Presentation Assessment", "Neoplasm Histologic Grade", "Disease Free Status", "Disease Free (Months)", "Subtype",
             "Cancer Type", "Cancer Type Detailed", "Number of Samples Per Patient", "Oncotree Code", "Sample Type", "Somatic Status",
             "TCGA PanCanAtlas Cancer Type Acronym", "Tumor Disease Anatomic Site", "Last Alive Less Initial Pathologic Diagnosis Date Calculated Day Value",
             "Informed consent verified", "Progression Free Status", "Progress Free Survival (Months)", "Prior Diagnosis", "Study ID", "Ethnicity Category",
             "American Joint Committee on Cancer Publication Version Type", "Informed consent verified", "In PanCan Pathway Analysis", "Tissue Source Site",
             "Tissue Source Site Code", "Form completion date", "Disease-specific Survival status", "Months of disease-specific survival",
             'Tissue Prospective Collection Indicator', 'Tissue Retrospective Collection Indicator']
```

```python
data = data.drop(drop_cols, axis=1)
```

```python
data.shape
```

```python
data.columns
```

# Numerical

```python
missing_cols = [col for col in data.columns if data[col].isna().sum() > 0]
missing_cols
```

```python
s = data[missing_cols].dtypes != 'object'
missing_num_cols = list(s[s].index)

missing_num_cols
```

```python
num_imputer = SimpleImputer(strategy='median')
```

```python
data[missing_num_cols] = num_imputer.fit_transform(data[missing_num_cols])
```

# Categorical
## Ordinal

```python
ID_cols = ['Patient ID', 'SAMPLE_ID']
```

```python
s = data.dtypes == 'object'
cat_cols_with_ids = list(s[s].index)
cat_cols = [col for col in cat_cols_with_ids if col not in ID_cols]
cat_cols
```

```python
# Get number of unique entries in each column with categorical data
object_nunique = list(map(lambda col: data[col].nunique(), cat_cols))
d = dict(zip(cat_cols, object_nunique))

# Print number of unique entries by column, in ascending order
sorted(d.items(), key=lambda x: x[1])
```

```python
# Too many values
too_many_unique_val_cols = ['Tumor Type', 'Other Patient ID']
```

```python
data = data.drop(too_many_unique_val_cols, axis=1)
```

```python
# Columns where order matters! (or just pure binary)
ordinal_cols = [col for col in data.columns if data[col].nunique() == 2]
ordinal_cols.extend(['American Joint Committee on Cancer Metastasis Stage Code', 'Neoplasm Disease Lymph Node Stage American Joint Committee on Cancer Code',
                     'International Classification of Diseases for Oncology, Third Edition ICD-O-3 Site Code', 'ICD-10 Classification',
                     'Neoplasm Disease Stage American Joint Committee on Cancer Code', 'American Joint Committee on Cancer Tumor Stage Code',
                     'International Classification of Diseases for Oncology, Third Edition ICD-O-3 Histology Code'])
```

```python
ordinal_encoder = OrdinalEncoder() 
cat_imputer = SimpleImputer(strategy='constant', fill_value=-1)
```

```python
data[ordinal_cols] = ordinal_encoder.fit_transform(data[ordinal_cols])
data[ordinal_cols] = cat_imputer.fit_transform(data[ordinal_cols])
```

## One-hot

```python
# Resetting index to concat with one-hot encoded data later
data = data.reset_index(drop=True)
```

```python
s = data.dtypes == 'object'

OH_cols = [col for col in list(s[s].index) if col not in ID_cols]
OH_cols
```

```python
data[OH_cols].isna().sum().sort_values()
```

```python
OH_imputer = SimpleImputer(strategy='constant')
OH_encoder = OneHotEncoder(sparse_output=False)
```

```python
data[OH_cols] = OH_imputer.fit_transform(data[OH_cols])
data[OH_cols]
```

```python
encoded_OH_data = pd.DataFrame(OH_encoder.fit_transform(data[OH_cols]))
encoded_OH_data.columns = OH_encoder.get_feature_names_out()
# print(encoded_OH_data.dtypes.to_string())
```

```python
data = data.drop(OH_cols, axis=1)
```

```python
data = pd.concat([data, encoded_OH_data], axis=1)
```

```python
demographics_cols = ['SAMPLE_ID',
    "Diagnosis Age",
    "Birth from Initial Pathologic Diagnosis Date",
    "Sex",
    "Race Category_American Indian or Alaska Native",
    "Race Category_Asian",
    "Race Category_Black or African American",
    "Race Category_White",
    "Race Category_missing_value",
    "Genetic Ancestry Label_AFR",
    "Genetic Ancestry Label_AFR_ADMIX",
    "Genetic Ancestry Label_AMR",
    "Genetic Ancestry Label_EAS",
    "Genetic Ancestry Label_EUR",
    "Genetic Ancestry Label_missing_value"
]

clinical_cols = ['SAMPLE_ID',
    "Neoplasm Disease Stage American Joint Committee on Cancer Code",
    "American Joint Committee on Cancer Metastasis Stage Code",
    "Neoplasm Disease Lymph Node Stage American Joint Committee on Cancer Code",
    "American Joint Committee on Cancer Tumor Stage Code",
    "ICD-10 Classification",
    "International Classification of Diseases for Oncology, Third Edition ICD-O-3 Histology Code",
    "International Classification of Diseases for Oncology, Third Edition ICD-O-3 Site Code",
    "Person Neoplasm Cancer Status",
    "New Neoplasm Event Post Initial Therapy Indicator",
    "Neoadjuvant Therapy Type Administered Prior To Resection Text",
    "Last Communication Contact from Initial Pathologic Diagnosis Date"
]

paraclinical_cols = ['SAMPLE_ID',
    "Aneuploidy Score",
    "Buffa Hypoxia Score",
    "Fraction Genome Altered",
    "MSI MANTIS Score",
    "MSIsensor Score",
    "Mutation Count",
    "Ragnum Hypoxia Score",
    "Winter Hypoxia Score",
    "Tumor Break Load",
    "TMB (nonsynonymous)"
]

treatment_cols = ['SAMPLE_ID', "Radiation Therapy"]

output_cols = ['SAMPLE_ID', "Overall Survival (Months)", "Overall Survival Status"]              
```

```python
luad_demographics = data[demographics_cols]
luad_clinical = data[clinical_cols]
luad_paraclinical = data[paraclinical_cols]
luad_treatment = data[treatment_cols]
luad_output = data[output_cols]
```

```python
data.to_csv('preprocessed_luad_non_omics.csv', index=False)
luad_demographics.to_csv('preprocessed_luad_demographics.csv', index=False)
luad_clinical.to_csv('preprocessed_luad_clinical.csv', index=False)
luad_paraclinical.to_csv('preprocessed_luad_paraclinical.csv', index=False)
luad_treatment.to_csv('preprocessed_luad_treatment.csv', index=False)
luad_output.to_csv('preprocessed_luad_output.csv', index=False)
```
